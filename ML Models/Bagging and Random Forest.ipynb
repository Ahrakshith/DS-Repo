```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('train.csv')
df.head()
```

```python
df.describe()
```

```python
num_cols=df.select_dtypes(include=['int64','float64','bool']).columns
print(num_cols)
```

```python
cols_to_include=['Id', 'MSSubClass', 'LotFrontage', 'LotArea', 'OverallQual',
       'OverallCond', 'YearBuilt', 'YearRemodAdd', 'MasVnrArea', 'BsmtFinSF1',
       'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', '1stFlrSF', '2ndFlrSF',
       'LowQualFinSF', 'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath',
       'HalfBath', 'BedroomAbvGr', 'KitchenAbvGr', 'TotRmsAbvGrd',
       'Fireplaces', 'GarageYrBlt', 'GarageCars', 'GarageArea', 'WoodDeckSF',
       'OpenPorchSF', 'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea',
       'MiscVal', 'MoSold', 'YrSold', 'SalePrice']
df=df[cols_to_include]
df.head()
```

```python
df.nunique().sort_values()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df['LotFrontage'].fillna(df['LotFrontage'].median(), inplace=True)
```

```python
df['MasVnrArea'].fillna(df['MasVnrArea'].mean(), inplace=True)

```

```python
df['GarageYrBlt'].fillna(0, inplace=True)
```

```python
print(df.isnull().sum().sum())
```

```python
df.isnull().sum()
```

```python
x=df.drop('SalePrice',axis=1)
```

```python
y=df['SalePrice']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,roc_auc_score,mean_squared_error,mean_absolute_error,r2_score
print(mean_squared_error(y_pred,y_test))
print(r2_score(y_pred,y_test))
```

```python
from sklearn.ensemble import BaggingRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error,r2_score
base_model=DecisionTreeRegressor(random_state=42)

bagging=BaggingRegressor(
    estimator=base_model,
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
bagging.fit(x_train,y_train)

y_predict=bagging.predict(x_test)

train_predict=bagging.predict(x_train)

mse=mean_squared_error(y_predict,y_test)

bias_train=np.sqrt(mean_squared_error(train_predict,y_train))

var_test=np.sqrt(mean_squared_error(y_predict,y_test))

print("mean squred error:",mse)

print("bias:",bias_train)

print("var_test:",var_test)
```
