```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
pd.set_option('display.max_columns',None)
```

```python
df=pd.read_csv('train.csv')
df.head()
```

```python
df.describe()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
cols_to_use =['Id','MSSubClass','MSZoning','LotFrontage','LotArea','Street','Alley','LotShape','YrSold','SalePrice']
df=df[cols_to_use]
df.head()
```

```python
df.isnull().sum()
```

```python
df.nunique()
```

```python
df['LotFrontage'].fillna(0)
```

```python
cols_to_fill_zero=['Id','MSSubClass','MSZoning','LotFrontage','LotArea']
df[cols_to_fill_zero]=df[cols_to_fill_zero].fillna(0)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = pd.get_dummies(df, drop_first=True)
```

```python
df.head()
```

```python
x=df.drop('SalePrice',axis=1)
```

```python
y=df['SalePrice']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,roc_auc_score,mean_squared_error,mean_absolute_error,r2_score
print(mean_squared_error(y_pred,y_test))
print(r2_score(y_pred,y_test))
```

```python
y_train_pred=lr.predict(x_train)
bias_train_rmse=np.sqrt(mean_squared_error(y_train,y_train_pred))

y_test_pred=lr.predict(x_test)
var_test_rmse=np.sqrt(mean_squared_error(y_test,y_test_pred))

print('bias' ,bias_train_rmse)
print('var', var_test_rmse)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
from sklearn.linear_model import Ridge
lr=Ridge()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,roc_auc_score,mean_squared_error,mean_absolute_error,r2_score
print(mean_squared_error(y_pred,y_test))
print(r2_score(y_pred,y_test))
```

```python
df1=pd.read_csv('train.csv')
df1.head()
```

```python
num_cols=df1.select_dtypes(include=['float64','int64','bool']).columns
print(num_cols)
```

```python
cols_to_include=['Id', 'MSSubClass', 'LotFrontage', 'LotArea', 'OverallQual',
       'OverallCond', 'YearBuilt', 'YearRemodAdd', 'MasVnrArea', 'BsmtFinSF1',
       'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', '1stFlrSF', '2ndFlrSF',
       'LowQualFinSF', 'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath',
       'HalfBath', 'BedroomAbvGr', 'KitchenAbvGr', 'TotRmsAbvGrd',
       'Fireplaces', 'GarageYrBlt', 'GarageCars', 'GarageArea', 'WoodDeckSF',
       'OpenPorchSF', 'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea',
       'MiscVal', 'MoSold', 'YrSold', 'SalePrice']
df1=df1[cols_to_include]
df1.head()
```

```python
df1.nunique()
```

```python
# See count of missing values per column
df1.isnull().sum()

# Or only show columns with NaN
df1.isnull().sum()[df1.isnull().sum() > 0]
```

```python
# Replace NaN with mean for specific columns
df1['MasVnrArea'].fillna(df1['MasVnrArea'].mean(), inplace=True)
df1['GarageYrBlt'].fillna(df1['GarageYrBlt'].mean(), inplace=True)

# Verify
print(df1[['MasVnrArea','GarageYrBlt']].isnull().sum())
```

```python
x1=df1.drop('SalePrice',axis=1)
```

```python
y1=df1['SalePrice']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x1,y1,test_size=0.2)
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score,recall_score,f1_score,roc_auc_score,mean_squared_error,mean_absolute_error,r2_score
print(mean_squared_error(y_pred,y_test))
print(r2_score(y_pred,y_test))
```

```python
y_train_pred=lr.predict(x_train)
bias_train_rmse=np.sqrt(mean_squared_error(y_train,y_train_pred))

y_test_pred=lr.predict(x_test)
var_test_rmse=np.sqrt(mean_squared_error(y_test,y_test_pred))

print('bias' ,bias_train_rmse)
print('var', var_test_rmse)
```
