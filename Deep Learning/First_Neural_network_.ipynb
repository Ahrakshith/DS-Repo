```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
url = "https://raw.githubusercontent.com/Ahrakshith/DS-Repo/main/datasets/Titanic-Dataset.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.describe()
```

```python
num_cols=df.select_dtypes(include=['int64','float64','bool']).columns.to_list()
cat_cols=df.select_dtypes(include=['object','category','bool']).columns.to_list()
print(num_cols)
print(cat_cols)
```

```python
df.isnull().sum()
```

```python
df.isna().any().any()
cols_with_na=df.columns[df.isna().any()].to_list()
print(cols_with_na)
```

```python
for c in num_cols:
    if df[c].isna().any():
        df[c].fillna(df[c].median(),inplace=True)

for c in cat_cols:
    if df[c].isna().any():
        df[c].fillna(df[c].mode()[0],inplace=True)
```

```python
df.isna().any().any()
df.isnull().sum()
```

```python
df['Title'] = df['Name'].str.extract(r',\s*([^\.]+)\.')[0].fillna('Unknown')

title_counts = df['Title'].value_counts()
rare_titles = title_counts[title_counts < 10].index
df['Title'] = df['Title'].replace(rare_titles, 'Other')


df['CabinDeck'] = df['Cabin'].fillna('Missing').astype(str).str[0]
df.loc[df['CabinDeck'] == 'M', 'CabinDeck'] = 'Missing'


df['FamilySize'] = df['SibSp'].fillna(0) + df['Parch'].fillna(0) + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)
```

```python
numeric_features = ['Age', 'Fare', 'FamilySize']


ordinal_features = ['Pclass']

categorical_nominal = ['Sex', 'Embarked', 'Title', 'CabinDeck']


drop_cols = ['PassengerId', 'Name', 'Ticket', 'Cabin']
df = df.drop(columns=[c for c in drop_cols if c in df.columns])


target_col = 'Survived'
```

```python
x = df[numeric_features + ordinal_features + categorical_nominal]
y = df[target_col]
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y,
                                                    test_size=0.2,
                                                    random_state=42,
                                                    stratify=y)
```

```python
x_train_encoded = pd.get_dummies(x_train, columns=categorical_nominal, drop_first=True)
x_test_encoded = pd.get_dummies(x_test, columns=categorical_nominal, drop_first=True)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(16, activation='relu', input_shape=(x_train_encoded.shape[1],)),
    Dense(8, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
history = model.fit(x_train_encoded, y_train, batch_size=15, validation_split=0.2, epochs=10)

print('model_accuracy:', model.evaluate(x_test_encoded.reindex(columns=x_train_encoded.columns, fill_value=0), y_test))
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train_encoded,y_train)
```

```python
y_pred = lr.predict(x_test_encoded.reindex(columns=x_train_encoded.columns, fill_value=0))
```

```python
# metrics_example.py — run this after you have y_test and y_pred
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# y_test = true labels
# y_pred = predicted labels from lr.predict(x_test)

acc  = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average='binary')  # for binary classification
rec  = recall_score(y_test, y_pred, average='binary')
f1   = f1_score(y_test, y_pred, average='binary')

print(f"Accuracy : {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"F1 score : {f1:.4f}\n")

# Confusion matrix (useful to inspect TP/FP/TN/FN)
cm = confusion_matrix(y_test, y_pred)
print("Confusion matrix:\n", cm)

# Nice text summary (handles multiclass automatically)
print("\nClassification report:\n")
print(classification_report(y_test, y_pred))
```
